```python
import numpy as np 
import pandas as pd

df = pd.read_csv("movies.csv")
df
```

```python
gen = df.iloc[0]["genres"].split("|")
gen
```

```python
tatCaGenres = set()

for genres in df["genres"]:
    for genre in genres.split("|"):
        tatCaGenres.add(genre)

print(tatCaGenres)
print(len(tatCaGenres))
```

```python
ten_phim_1 = "Black Butler: Book of the Atlantic"
phim_1 = df[df["title"].str.contains(ten_phim_1, case=False, na=False, regex=False)]
the_loai_1 = set(phim_1.iloc[0]["genres"].split("|"))

ten_phim_2 = "No Game No Life: Zero"
phim_2 = df[df["title"].str.contains(ten_phim_2, case=False, na=False, regex=False)]
the_loai_2 = set(phim_2.iloc[0]["genres"].split("|"))

bang = pd.DataFrame(columns=["The loai phim 1  ", "Phim 2 co khong?  ", "Gia tri"])
vector = []

for i in the_loai_1:
    if i in the_loai_2:
        bang.loc[len(bang)] = [i + "  ", "Yes  ", 1]
        vector.append(1)
    else:
        bang.loc[len(bang)] = [i + "  ", "No  ", 0]
        vector.append(0)

print(bang)
```

```python
the_loai = ["Action", "Comedy", "Drama"]

ten_phim_1 = "Black Butler: Book of the Atlantic"
phim_1 = df[df["title"].str.contains(ten_phim_1, case=False, na=False, regex=False)]
the_loai_1 = set(phim_1.iloc[0]["genres"].split("|"))

ten_phim_2 = "No Game No Life: Zero"
phim_2 = df[df["title"].str.contains(ten_phim_2, case=False, na=False, regex=False)]
the_loai_2 = set(phim_2.iloc[0]["genres"].split("|"))

bang = pd.DataFrame(columns=["Doi tuong"] + the_loai)
bang

v1 = []
v2 = []
for i in the_loai:
    if i in the_loai_1:
        v1.append(1)
    else:
        v1.append(0)
    if i in the_loai_2:
        v2.append(1)
    else:
        v2.append(0)

bang.loc[len(bang)] = [ten_phim_1] + v1
bang.loc[len(bang)] = [ten_phim_2] + v2

dot = np.dot(v1, v2)
l2_1 = np.sqrt(np.sum(np.array(v1)**2))
l2_2 = np.sqrt(np.sum(np.array(v2)**2))
cosine = dot / (l2_1 * l2_2)

print(bang)
print()
print("dot =", dot)
print("l2_1 =", l2_1)
print("l2_2 =", l2_2)
print("cosine =", cosine)
```

```python
tl = ["Action", "Horror", "Comedy"]

ten_phim = "Black Butler: Book of the Atlantic"
phim = df[df["title"].str.contains(ten_phim, case=False, na=False, regex=False)]
the_loai = set(phim.iloc[0]["genres"].split("|"))

u = []
for i in tl:
    if i in the_loai:
        u.append(1)
    else:
        u.append(0)

bang = pd.DataFrame(columns=["Phim", "Vector", "Dot", "Cosine"])
bang

for _, i in df.head(10).iterrows():
    v = []
    for j in tl:
        if j in i["genres"].split("|"):
            v.append(1)
        else:
            v.append(0)
    dot = np.dot(u, v)
    bang.loc[len(bang)] = [i["title"], v, dot, dot / (np.linalg.norm(u) * np.linalg.norm(v))]

bang
```
