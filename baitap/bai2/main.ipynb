```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

df = pd.read_csv("SMSSpamCollection.csv", sep="\t", header=None, names=["label", "message"])
df
```

```python
def count_char(message):
    letter_count = 0
    digit_count = 0
    for char in message:
        if char.isalpha():
            letter_count += 1
        elif char.isdigit():
            digit_count += 1
    return pd.Series([letter_count, digit_count])
```

```python
df[["letter_count", "digit_count"]] = (
    df["message"].apply(count_char))
```

```python
X = df[["letter_count", "digit_count"]]
y = df["label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,
    test_size=0.2,
    random_state=36,
    stratify=y
)
```

```python
knn = KNeighborsClassifier(n_neighbors=5, metric="euclidean")
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
for k in range(1, 21, 2):
    knn = KNeighborsClassifier(n_neighbors=k, metric="euclidean")
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"K = {k}, Accuracy = {accuracy:.4f}")
```
